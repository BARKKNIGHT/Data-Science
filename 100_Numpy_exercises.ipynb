#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)

````python
mat6 = np.zeros((5,5))
accepted = [mat6.shape[0]-1,0]
for i in range(len(mat6)):
    for j in range(len(mat6[0])):
        if i in accepted or j in accepted:
            mat6[i][j] = 1
mat6
````

#### 29. How to round away from zero a float array ? (★☆☆)

````python
float_array = np.array([1.2, -1.2, 2.5, -2.5, 3.7, -3.7])

rounded_array = np.array([np.ceil(i) if i > 0 else np.floor(i) for i in float_array])

print("Original array:", float_array)
print("Rounded array:", rounded_array)
````

#### 36. Extract the integer part of a random array of positive numbers using 4 different methods (★★☆)

````python
random_array = np.random.random(10) * 100
print("Original array:", random_array)

integer_part_floor = np.floor(random_array)
print("Integer part (using np.floor):", integer_part_floor)

integer_part_trunc = np.trunc(random_array)
print("Integer part (using np.trunc):", integer_part_trunc)

integer_part_astype = random_array.astype(int)
print("Integer part (using astype(int)):", integer_part_astype)

integer_part_round = np.round(random_array, decimals=0).astype(int)
print("Integer part (using np.round):", integer_part_round)
````

#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

````python
cartesian = np.random.random((10, 2)) * 10
print("Cartesian coordinates (x, y):")
print(cartesian)

x = cartesian[:, 0]
y = cartesian[:, 1]

r = np.sqrt(x**2 + y**2)
theta = np.arctan2(y, x)

polar = np.column_stack((r, theta))
print("\nPolar coordinates (r, theta):")
print(polar)
````

#### 46. Create a structured array with `x` and `y` coordinates covering the [0,1]x[0,1] area (★★☆)

````python
grid_size = 10

x = np.linspace(0, 1, grid_size)
y = np.linspace(0, 1, grid_size)

xx, yy = np.meshgrid(x, y)

structured_array = np.zeros(xx.shape, dtype=[('x', float), ('y', float)])
structured_array['x'] = xx
structured_array['y'] = yy

print("Structured array:")
print(structured_array)
````

#### 48. Print the minimum and maximum representable value for each numpy scalar type (★★☆)

````python
print("Integer types:")
for dtype in [np.int8, np.int16, np.int32, np.int64, np.uint8, np.uint16, np.uint32, np.uint64]:
    info = np.iinfo(dtype)
    print(f"{dtype.__name__}: min={info.min}, max={info.max}")

print("\nFloating-point types:")
for dtype in [np.float16, np.float32, np.float64]:
    info = np.finfo(dtype)
    print(f"{dtype.__name__}: min={info.min}, max={info.max}, eps={info.eps}")
````

#### 50. How to find the closest value (to a given scalar) in a vector? (★★☆)

````python
vector = np.array([1, 3, 8, 10, 15, 20])
scalar = 12

closest_index = np.argmin(np.abs(vector - scalar))

closest_value = vector[closest_index]

print("Vector:", vector)
print("Scalar:", scalar)
print("Closest value:", closest_value)
````
