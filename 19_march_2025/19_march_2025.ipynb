Suppose the price of the phone is 40000. Calculate the prices of this phone after providing 5%, 10%, 15%, and 20% discounts respectively.

```python

def discount_cal(price):
    return [price - (price*i)/100 for i in [5,10,15,20]]
```

```python
discount_cal(10000)
```

```python
def discount_calc(price, discount):
    return price - ((price*discount)/100)
```

```python
discount_calc(10000,5)
```

```python
def is_even(num):
    return num%2 == 0
```

```python
is_even(2)
```

```python
def factorial(num):
    if num == 1 or num == 0:
        return 1
    else:
        return num*factorial(num-1)
```

```python
factorial(5)
```
