# W.A.P in Python that prints whether the customer can buy the phone based on his budget


```python
budget = int(input("Enter your budget:"))
```

```python
if budget >= 30000:
    print('You can buy the phone!')
else:
    print("You can't buy the phone!")
```

# Q2

```python
price_map = {
    32:15000,
    64:20000,
    128:30000,
    256:40000
}
```

```python
storage_required = int(input('Enter desired phone storage:'))
```

```python
if storage_required in price_map.keys():
    print(f'Your phone will cost:{price_map[storage_required]}')
else:
    print('Invalid storage!')
```

# Q3

```python
round_threshold = {
    1:80,
    2:70,
    3:60
}
```

```python
score1 = int(input('Enter your score in round 1:'))
```

```python
if score1 > round_threshold[1]:
    score2 = int(input('Enter your score in round 2:'))
    if score2 > round_threshold[2]:
        score3 = int(input('Enter your score in round 3:'))
        if score3 > round_threshold[3]:
            print('You passed!')
        else:
            print('You failed!')
    else:
        print("You can't go to round 3!")
else:
    print("You can't go to round 2!")
```

```python
flag = True
```

```python
while flag:
    username = input('Enter your username:')
    if username == 'admin':
        while flag:
            password = input('Enter your password:')
            if password == '1234':
                print('Logged in!')
                flag = False
            else:
                print('Incorrect password!')
    else:
        print('Incorrect username!')
```
