```python
import pymysql.cursors
import pandas as pd
```

```python
def get_connection():
    connection = pymysql.connect(host='localhost',
                                 user='root',
                                 password='1234',
                                 database='hr',
                                 charset='utf8mb4',
                                 cursorclass=pymysql.cursors.DictCursor)
    return connection
```

# 1. Retrieve all details of employees. 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = "Select * from employees"
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 7. Retrieve employees with job titles containing the word ‘Manager.’ 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT e.*, j.job_title 
                FROM employees e
                JOIN jobs j ON e.job_id = j.job_id
                WHERE j.job_title LIKE '%Manager%';
        """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 14. Convert the hire_date of employees to display only the year.

# 15. Retrieve the minimum and maximum salary for each job title. 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT j.job_title, MIN(e.salary) AS min_salary, MAX(e.salary) AS max_salary
                FROM employees e
                JOIN jobs j ON e.job_id = j.job_id
                GROUP BY j.job_title
              """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 17. List the employees along with their job titles and the location of their department. 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT e.first_name, e.last_name, j.job_title, l.city, l.country_id
                FROM employees e
                JOIN jobs j ON e.job_id = j.job_id
                JOIN departments d ON e.department_id = d.department_id
                JOIN locations l ON d.location_id = l.location_id;
              """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 18. Retrieve the department names along with the count of employees in each department. 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT d.department_name, COUNT(e.employee_id) AS employee_count
                 FROM departments d
                 LEFT JOIN employees e ON d.department_id = e.department_id
                 GROUP BY d.department_name;
              """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 19. Find employees who have the same job as their manager. 

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT e.first_name, e.last_name, j.job_title
                 FROM employees e
                 JOIN employees m ON e.manager_id = m.employee_id
                 JOIN jobs j ON e.job_id = j.job_id
                 WHERE e.job_id = m.job_id;
              """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```

# 20. Display the names of employees who worked in different jobs in the past (use job_history).

```python
connection = get_connection()
with connection:
    with connection.cursor() as cursor:
        sql = """SELECT DISTINCT e.first_name, e.last_name
                FROM employees e
                JOIN job_history jh ON e.employee_id = jh.employee_id
                WHERE e.job_id != jh.job_id;
              """
        cursor.execute(sql)
        data = cursor.fetchall()
        df = pd.DataFrame(data)
        print(df.head())
```
