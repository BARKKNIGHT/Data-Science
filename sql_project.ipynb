```python
import pymysql.cursors
import pandas as pd
```

```python
def get_connection():
    connection = pymysql.connect(host='localhost',
                                 user='root',
                                 password='1234',
                                 database='hr',
                                 charset='utf8mb4',
                                 cursorclass=pymysql.cursors.DictCursor)
    return connection
```

```python
def run_query(query):
    connection = get_connection()
    with connection:
        with connection.cursor() as cursor:
            sql = query
            cursor.execute(sql)
            data = cursor.fetchall()
            df = pd.DataFrame(data)
            return df
```

# 1. Retrieve all details of employees. 

```python
run_query("Select * from employees")
```

# 2. Display the first name, last name, and email of all employees. 

```python
run_query("SELECT first_name, last_name, email FROM employees")
```

# 3. Retrieve the distinct job titles from the jobs table.

```python
run_query("SELECT DISTINCT job_title FROM jobs")
```

# 4. Find the total number of employees in the company.

```python
run_query("SELECT COUNT(*) AS total_employees FROM employees")
```

# 5. Retrieve the employees who were hired after January 1, 2015.

```python
run_query("SELECT * FROM employees WHERE hire_date > '2015-01-01'")
```

# 6. List all employees who have a salary greater than 5000. 

```python
run_query("SELECT * FROM employees WHERE salary > 5000")
```

# 7. Retrieve employees with job titles containing the word ‘Manager.’ 

```python
run_query("SELECT e.*, j.job_title FROM employees e JOIN jobs j ON e.job_id = j.job_id WHERE j.job_title LIKE '%Manager%'")
```

# 8. Retrieve all employees whose first name starts with ‘A’ and ends with ‘n.’ 

```python
run_query("SELECT * FROM employees WHERE first_name LIKE 'A%n'")
```

# 9. Display the employees who do not have a commission. 

```python
run_query("SELECT * FROM employees WHERE commission_pct IS NULL OR commission_pct = 0")
```

# 10. Retrieve the top 5 highest-paid employees. 

```python
run_query("SELECT * FROM employees ORDER BY salary DESC LIMIT 5")
```

# 11. Find the average salary of all employees.

```python
run_query("SELECT AVG(salary) AS average_salary FROM employees")
```

# 12. Retrieve the total number of employees working in each department. 

```python
run_query("SELECT department_id, COUNT(*) AS employee_count FROM employees GROUP BY department_id")
```

# 13. Display the employee's first name and the length of their first name. 

```python
run_query("SELECT first_name, LENGTH(first_name) AS name_length FROM employees")
```

# 14. Convert the hire_date of employees to display only the year. 

```python
run_query("SELECT first_name, last_name, EXTRACT(YEAR FROM hire_date) AS hire_year FROM employees")
```

# 15. Retrieve the minimum and maximum salary for each job title. 

```python
run_query("SELECT j.job_title, MIN(e.salary) AS min_salary, MAX(e.salary) AS max_salary FROM employees e JOIN jobs j ON e.job_id = j.job_id GROUP BY j.job_title")
```

# 16. Retrieve the employee names along with their department names. 

```python
run_query("SELECT e.first_name, e.last_name, d.department_name FROM employees e JOIN departments d ON e.department_id = d.department_id;")
```

# 17. List the employees along with their job titles and the location of their department. 

```python
run_query("SELECT e.first_name, e.last_name, j.job_title, l.city, l.country_id FROM employees e JOIN jobs j ON e.job_id = j.job_id JOIN departments d ON e.department_id = d.department_id JOIN locations l ON d.location_id = l.location_id")
```

# 18. Retrieve the department names along with the count of employees in each department. 

```python
run_query("SELECT d.department_name, COUNT(e.employee_id) AS employee_count FROM departments d LEFT JOIN employees e ON d.department_id = e.department_id GROUP BY d.department_name")
```

# 19. Find employees who have the same job as their manager. 

```python
run_query("SELECT e.first_name, e.last_name, j.job_title FROM employees e JOIN employees m ON e.manager_id = m.employee_id JOIN jobs j ON e.job_id = j.job_id WHERE e.job_id = m.job_id")
```

# 20. Display the names of employees who worked in different jobs in the past (use job_history).

```python
run_query("SELECT DISTINCT e.first_name, e.last_name FROM employees e JOIN job_history jh ON e.employee_id = jh.employee_id WHERE e.job_id != jh.job_id")
```
